```python
import math

from optionslib.vol.implied import (
    DEFAULT_CONFIG,
    NewtonExit,
    newton_method,
    normalized_price,
)

cfg = DEFAULT_CONFIG


def guess_linear(x, s_c, b_target):
    return b_target * math.sqrt(2.0 * math.pi)


def guess_above_sc(x, s_c, b_target):
    return s_c * 1.5


def guess_hybrid(x, s_c, b_target):
    if abs(x) < 0.5:
        return b_target * math.sqrt(2.0 * math.pi)
    return s_c * 1.5


for x in [-1.5, -1.0, -0.5, -0.2, 0.0, 0.2, 0.5, 1.0, 1.5]:
    for s_true in [0.1, 0.5, 1.0, 2.0, 4.0]:
        b_target = normalized_price(x, 1, s_true)
        s_c = math.sqrt(2.0 * abs(x))

        for name, g in [
            ("lin", guess_linear),
            ("1.5sc", guess_above_sc),
            ("hybrid", guess_hybrid),
        ]:
            s0 = min(max(g(x, s_c, b_target), cfg.s_lo), cfg.s_hi)
            r = newton_method(cfg, x, 1, b_target, s0)
            ok = r.exit_reason == NewtonExit.CONVERGED
            err = abs(r.s - s_true) if ok else float("nan")
            print(
                f"x={x:+.2f} s_true={s_true:.2f} {name:6s} "
                f"s0={s0:.3f} {r.exit_reason:15s} iters={r.iterations} err={err:.2e}"
            )
```
