**Question:** This check asks whether initial_guess needs a small-|x| threshold or whether x==0.0 is sufficient.

**Setup:** At x=0, s_c=sqrt(2|x|) will be zero and normalized_price rejects s_c<=0. The concern was with small but nonzero x where s_c is tiny and the evaluation degrades. We measure normalized price (b) and vega at s = s_c at x from 1e-12 to 1e-2. 

**Findings:**
1. In each case vega converges to 1/sqrt(2*pi) rather than underflowing. At s=s_c the exponent x^2/(2*s_c^2) reduces to |x|/4, which vanishes with x. So nothing degrades in this case. 
2. b/s_c=0.3989 result holds throughout. Small-s linearization b=s/sqrt(2*pi) holds across the whole range from 1e-12 to 1e-2. x=0 branch converges continuously to s_c branches as x->0. 
    
**Conclusion:** Hence exact equality in initial_guess is safe. The branch exists only for literal x=0.0, reachable when a parity-regressed forward lands exactly on a listed strike.
